```python
class Perceptron:
    def __init__(self, n_inputs, learning_rate = 0.01, epochs = 100):
        self.lr = learning_rate
        self.weights = [0.0] * n_inputs
        self.b = 0.0
        self.epochs = epochs

    def predict(self, inputs):
        total = sum(w * x for w, x in zip(self.weights, inputs))
        total += self.b
        return 1 if total >= 0 else 0

    def train(self, training_data):
        for epoch in range(self.epochs):
            for inputs, target in training_data:
                prediction = self.predict(inputs)
                error = prediction - target
                if error != 0:
                    for i in range(len(self.weights)):
                        self.weights[i] -= self.lr * error * inputs[i]
                    self.b -= self.lr * error
                
```

```python
AND_GATE = [
    ([0,0], 0),
    ([1,0], 0),
    ([0,1], 0),
    ([1,1], 1)
]

OR_GATE = [
    ([0,0], 0),
    ([1,0], 1),
    ([0,1], 1),
    ([1,1], 1)
]

print("Perceptron for AND gate")
AND = Perceptron(n_inputs=2)
AND.train(AND_GATE)
print(f"Prediction for (0, 1): {AND.predict((0,1))}")

print(f"Weights: {AND.weights}")
print(f"Bias: {AND.b}")

print("\n\nPerceptron for OR gate")
OR = Perceptron(n_inputs=2)
OR.train(OR_GATE)
print(f"Prediction for (0, 1): {OR.predict((0,1))}")

print(f"Weights: {OR.weights}")
print(f"Bias: {OR.b}")
```
